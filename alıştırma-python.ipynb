```python
# 1. Ekrana Yazı Yazdırma (print)

# Ekrana "Python öğreniyorum" yazdıran bir kod yazın.
print("Python öğreniyorum")

# Adınızı ve soyadınızı aynı print satırında yazdırın.
print("Sena Yetim")

# Ekrana önce bir sayı, sonra bir yazı yazdıran iki ayrı print satırı yazın.
print(14)
print("Bolu")
```

```python
# 2. Yorum Satırları

# Bir satır kod yazın ve yanına o satırın ne işe yaradığını açıklayan bir yorum ekleyin.
print("Merhaba Dünya")  # Ekrana Merhaba Dünya metnini yazdırır.

# Çok satırlı bir yorum kullanarak adınızı, yaşınızı ve şehrinizi not edin (kod çalıştırılmadan).
"""
Ad: Sena
Yaş: 25
Şehir: İstanbul
"""
```

```python
# 3. Değişkenler

# sehir adında bir değişken oluşturup içine yaşadığınız şehri atayın ve yazdırın.
sehir = "İstanbul"
print(sehir)

# puan adında bir değişkene 50 değerini verin, sonra değerini 75 olarak güncelleyip her iki değeri de yazdırın.
puan = 50
print(puan)
puan = 75
print(puan)
```

```python
# 4. Veri Tipleri

# Bir tam sayı, bir ondalıklı sayı, bir yazı ve bir boolean değişken oluşturup her birinin tipini type() ile yazdırın.
tam_sayi = 10
ondalikli = 4.5
yazi_degeri = "Merhabalar"
aktif_mi = True

print(type(tam_sayi))
print(type(ondalikli))
print(type(yazi_degeri))
print(type(aktif_mi))

# 10 ve "10" değerlerinin tiplerini karşılaştırın — ikisi de aynı mı?
print(type(10) == type("10"))  # False döner, çünkü biri int diğeri str türündedir.
```

```python
# 5. Temel Matematik İşlemleri

# 15 ve 4 sayılarını kullanarak toplama, çıkarma, çarpma ve bölme işlemlerini yapın.
print(15 + 4)
print(15 - 4)
print(15 * 4)
print(15 / 4)

# 17'nin 5'e bölümünden kalanı (mod) bulun.
print(17 % 5)

# 2'nin 8. kuvvetini (2 üzeri 8) hesaplayın.
print(2 ** 8)
```

```python
# 6. Kullanıcıdan Veri Alma (input)

# Kullanıcıdan adını alıp "Merhaba, [ad]!" şeklinde yazdıran bir program yazın.
ad = input("Adınız nedir? ")
print(f"Merhaba, {ad}!")

# Kullanıcıdan yaşını alıp 10 yıl sonraki yaşını hesaplayıp yazdıran bir program yazın.
yas = int(input("Yaşınız kaç? "))
print(f"10 yıl sonra {yas + 10} yaşında olacaksınız.")
```

```python
# 7. Karşılaştırma Operatörleri

# a = 8, b = 12 için a > b, a < b ve a == b sonuçlarını yazdırın.
a = 8
b = 12
print(a > b)
print(a < b)
print(a == b)

# Kullanıcıdan bir sayı alıp bu sayının 100'e eşit olup olmadığını kontrol edin.
sayi_kontrol = int(input("Bir sayı giriniz: "))
print(sayi_kontrol == 100)
```

```python
# 8. Mantıksal Operatörler

# yas = 22 ve ogrenci_mi = False değişkenleri için and, or ve not sonuçlarını yazdırın.
yas = 22
ogrenci_mi = False
print(yas >= 18 and ogrenci_mi)
print(yas >= 18 or ogrenci_mi)
print(not ogrenci_mi)

# Bir kişinin hem 18 yaşından büyük hem de kayıtlı olup olmadığını kontrol eden bir ifade yazın.
kisi_yas = 20
kayitli_mi = True
print(kisi_yas > 18 and kayitli_mi)
```

```python
# 9. Basit if-else Kullanımı

# Kullanıcıdan bir sayı alıp bu sayının pozitif mi negatif mi olduğunu yazdıran bir program yazın.
sayi_input = int(input("Bir sayı girin: "))
if sayi_input > 0:
    print("Pozitif sayı")
elif sayi_input < 0:
    print("Negatif sayı")
else:
    print("Sayı sıfırdır")

# Kullanıcıdan not alıp 50 ve üzeri ise "Geçti", altında ise "Kaldı" yazdıran bir program yazın.
not_degeri = int(input("Notunuzu girin: "))
if not_degeri >= 50:
    print("Geçti")
else:
    print("Kaldı")
```

```python
# 10. String (Yazı) ile Temel İşlemler

# Adınızı ve soyadınızı ayrı değişkenlerde tutup birleştirerek tam adınızı yazdırın.
isim = "Sena"
soyisim = "Yetim"
tam_isim = isim + " " + soyisim
print(tam_isim)

# Bir kelimenin kaç harften oluştuğunu len() ile bulun.
kelime = "Bilgisayar"
print(len(kelime))

# Bir yazıyı tamamen büyük harfe, sonra tamamen küçük harfe çevirin.
ornek_metin = "Sena Yetim"
print(ornek_metin.upper())
print(ornek_metin.lower())
```

```python
# 11. Basit Bir Program Örneği

# Kullanıcıdan iki sayı alıp çarpımını yazdıran bir program yazın.
sayi1 = int(input("1. sayıyı girin: "))
sayi2 = int(input("2. sayıyı girin: "))
print(f"Sayıların Çarpımı: {sayi1 * sayi2}")

# Kullanıcıdan bir sayı alıp bu sayının karesini (kendisiyle çarpımını) hesaplayan bir program yazın.
kare_sayisi = int(input("Karesi alınacak sayıyı girin: "))
print(f"Sayının Karesi: {kare_sayisi ** 2}")
```
