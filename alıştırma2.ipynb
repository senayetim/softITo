```python
# 1. Değişkenlerle Basit Hesaplamalar

# Birim fiyat ve adet ile toplam tutar
birim_fiyat = 45.5
adet = 4
toplam_tutar = birim_fiyat * adet
print(f"Toplam Tutar: {toplam_tutar} TL")

# Maaş ve zam hesaplama
maas = 8000
maas += 1500  # maas = maas + 1500 ile aynı şeydir
print(f"Yeni Maaş: {maas} TL")
```

```python
# 2. İç İçe (Nested) if-else Kullanımı

# Yaş ve öğrenci durumu kontrolü
yas = 22
ogrenci_mi = True

if yas > 65:
    print("Ücretsiz")
else:
    if ogrenci_mi:
        print("İndirimli")
    else:
        print("Tam ücret")

# Sayının hem pozitif hem çift olup olmadığını iç içe if ile kontrol
sayi = 16
if sayi > 0:
    if sayi % 2 == 0:
        print("Sayı pozitif ve çift.")
    else:
        print("Sayı pozitif ama tek.")
else:
        print("Sayı pozitif değil.")
```

```python
# 3. elif ile Çoklu Koşullar


# Sıcaklık durumları
sicaklik = 24
if sicaklik > 30:
    print("Sıcak")
elif 15 <= sicaklik <= 30:
    print("Ilıman")
else:
    print("Soğuk")

# Mevsim bulma (Ay numarasına göre)
ay = 10
if ay in [12, 1, 2]:
    print("Kış")
elif ay in [3, 4, 5]:
    print("İlkbahar")
elif ay in [6, 7, 8]:
    print("Yaz")
elif ay in [9, 10, 11]:
    print("Sonbahar")
else:
    print("Geçersiz ay numarası girdiniz.")
```

```python
# 4. for Döngüsü ile Liste Üzerinde İşlem

sayilar = [5, 15, 8, 22, 3, 19]

# max() fonksiyonu kullanmadan en büyük sayıyı bulma
en_buyuk = sayilar[0]
for s in sayilar:
    if s > en_buyuk:
        en_buyuk = s
print(f"En büyük sayı: {en_buyuk}")

# 10'dan büyük olanların adedini bulma
adet_10_buyuk = 0
for s in sayilar:
    if s > 10:
        adet_10_buyuk += 1
print(f"10'dan büyük sayı adedi: {adet_10_buyuk}")
```

```python
# 5. while Döngüsü ile Basit Sayaç

# 10'dan 1'e geri sayım
i = 10
while i >= 1:
    print(i)
    i -= 1

# Doğru şifre girene kadar soran döngü
sifre = ""
while sifre != "python123":
    sifre = input("Şifreyi giriniz: ")
    if sifre != "python123":
        print("Hatalı şifre, tekrar deneyin.")
print("Şifre doğru, giriş yapıldı!")
```

```python
# 6. Döngü İçinde Koşul Kullanma

# 1'den 20'ye kadar 3'e tam bölünenler
for sayi in range(1, 21):
    if sayi % 3 == 0:
        print(sayi)

# Listeden pozitif ve negatifleri ayırma
liste = [-4, 12, 5, -8, 0, 23, -1]
pozitifler = []
negatifler = []

for s in liste:
    if s > 0:
        pozitifler.append(s)
    elif s < 0:
        negatifler.append(s)

print(f"Pozitifler: {pozitifler}")
print(f"Negatifler: {negatifler}")
```

```python
# 7. Basit Fonksiyon Tanımlama

# Bir sayının küpünü alan fonksiyon
def kup_al(sayi):
    return sayi ** 3

# İki sayının toplamını döndüren fonksiyon
def toplamini_bul(sayi1, sayi2):
    return sayi1 + sayi2

print("3'ün küpü:", kup_al(3))
print("Toplam:", toplamini_bul(5, 10))
```

```python
# 8. Fonksiyon + Koşul Birlikte

# Sayının pozitif, negatif veya sıfır olduğunu döndüren fonksiyon
def sayi_durumu(n):
    if n > 0:
        return "Pozitif"
    elif n < 0:
        return "Negatif"
    else:
        return "Sıfır"

# Üç sayı arasından en büyüğünü bulan fonksiyon
def en_buyugu_bul(a, b, c):
    return max(a, b, c)

print(sayi_durumu(-5))
print("En büyük sayı:", en_buyugu_bul(12, 45, 23))
```

```python
# 9. Liste Oluşturma ve İşleme

# Boş listeye kullanıcıdan alınan 5 şehri ekleme ve numaralandırma
sehirler = []
for i in range(5):
    sehir = input(f"{i+1}. şehri giriniz: ")
    sehirler.append(sehir)

print("\nNumaralandırılmış Şehirler:")
for idx, s in enumerate(sehirler):
    print(f"{idx+1}. {s}")

# remove() ile listeden eleman silme
silinecek = input("\nListeden silmek istediğiniz şehri yazın: ")
if silinecek in sehirler:
    sehirler.remove(silinecek)
    print("Güncel liste:", sehirler)
else:
    print("Bu şehir listede yok.")
```

```python
# 10. Kullanıcıdan Birden Fazla Veri Alma

# Ad, soyad alıp 2026 yılına göre yaş hesaplama
ad = input("Adınız: ")
soyad = input("Soyadınız: ")
dogum_yili = int(input("Doğum yılınız: "))
yas = 2026 - dogum_yili
print(f"Sayın {ad} {soyad}, yaşınız: {yas}")

# Boy ve kilo ile BMI (Vücut Kitle İndeksi) hesaplama
boy_cm = float(input("Boyunuzu cm girin (örn: 170): "))
kilo = float(input("Kilonuzu kg girin (örn: 65): "))
bmi = kilo / ((boy_cm / 100) ** 2)
print(f"Vücut Kitle İndeksiniz (BMI): {bmi:.2f}")
```

```python
# 11. Küçük Uygulama — Not Ortalaması Hesaplama

notlar = []
for i in range(4):
    n = float(input(f"{i+1}. dersin notunu giriniz: "))
    notlar.append(n)

ortalama = sum(notlar) / len(notlar)
print(f"\nOrtalamanız: {ortalama:.2f}")

# Harf notu dönüşümü
if ortalama >= 85:
    print("Harf Notu: AA")
elif ortalama >= 70:
    print("Harf Notu: BB")
elif ortalama >= 50:
    print("Harf Notu: CC")
else:
    print("Harf Notu: FF (Kaldınız)")

print(f"En yüksek not: {max(notlar)}")
print(f"En düşük not: {min(notlar)}")
```
