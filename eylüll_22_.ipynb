```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report
```

```python
df = pd.read_csv('MathE dataset (4).csv', sep=';', encoding='latin1')
```

```python
# 2. Eksik Veri Kontrolü 

missing_values = df.isnull().sum()
print(missing_values)
```

```python
# 3. REGRESYON MODELLERİ (Linear & Polynomial Regression)

X_reg = df[['Student Country', 'Question Level', 'Topic', 'Subtopic']]
y_reg = df['Question ID']

# Kategorik değişkenleri sayısallaştırma (One-Hot Encoding)
X_reg_encoded = pd.get_dummies(X_reg, drop_first=True)

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg_encoded, y_reg, test_size=0.2, random_state=42)
```

```python
# A) Linear Regression
lin_reg = LinearRegression()
lin_reg.fit(X_train_r, y_train_r)
y_pred_lin = lin_reg.predict(X_test_r)

lin_r2 = r2_score(y_test_r, y_pred_lin)
lin_rmse = np.sqrt(mean_squared_error(y_test_r, y_pred_lin))

print(f"Linear Regression -> R2 Score: {lin_r2:.4f} | RMSE: {lin_rmse:.4f}")
```

```python
# B) Polynomial Regression (Degree = 2)
poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_poly = poly.fit_transform(X_train_r)
X_test_poly = poly.transform(X_test_r)

poly_reg = LinearRegression()
poly_reg.fit(X_train_poly, y_train_r)
y_pred_poly = poly_reg.predict(X_test_poly)

poly_r2 = r2_score(y_test_r, y_pred_poly)
poly_rmse = np.sqrt(mean_squared_error(y_test_r, y_pred_poly))

print(f"Polynomial Regression -> R2 Score: {poly_r2:.4f} | RMSE: {poly_rmse:.4f}\n")
```

```python
# 4. SINIFLANDIRMA MODELI (Logistic Regression)

X_clf = df[['Student Country', 'Question Level', 'Topic', 'Subtopic']]
y_clf = df['Type of Answer']

X_clf_encoded = pd.get_dummies(X_clf, drop_first=True)
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_clf_encoded, y_clf, test_size=0.2, random_state=42)

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_c, y_train_c)
y_pred_log = log_reg.predict(X_test_c)

acc = accuracy_score(y_test_c, y_pred_log)
print(f"Logistic Regression Accuracy (Dogruluk Orani): {acc:.4f}\n")

report_dict = classification_report(y_test_c, y_pred_log, output_dict=True)
report_df = pd.DataFrame(report_dict).transpose()

display(report_df.round(4))
```
